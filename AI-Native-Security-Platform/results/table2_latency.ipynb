# Table II: End-to-End Latency Breakdown Analysis
This notebook reproduces Table II and Figure 1 for the paper **An AI-Native Event-Driven Edge-Cloud Architecture for Autonomous Network Security**.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('table2_latency.csv')
display(df)

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.barh(df['Pipeline Stage'][:-1], df['Avg Latency (ms)'][:-1].astype(float), color='#2b5c8f')
ax.set_xlabel('Latency (ms)')
ax.set_title('Pipeline Stage Latency Breakdown')
plt.tight_layout()
plt.show()
```
