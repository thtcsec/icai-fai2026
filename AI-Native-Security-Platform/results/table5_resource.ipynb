# Table V: Edge CPU & RAM Scaling Analysis
This notebook reproduces Table V and Figure 4 comparing AI-Native Edge Gateway vs Legacy Inline DPI Proxy under scaling throughput.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('table5_resource.csv')
display(df)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3.5))
ax1.plot(df['Throughput (events/s)'], df['AI Edge CPU (%)'].str.rstrip('%').astype(float), 'o-', label='AI Edge')
ax1.plot(df['Throughput (events/s)'], df['Legacy DPI CPU (%)'].str.rstrip('%').astype(float), 's--', label='DPI Proxy')
ax1.set_title('CPU Overhead')
ax1.legend()

ax2.plot(df['Throughput (events/s)'], df['AI Edge RAM (MB)'].str.rstrip(' MB').astype(float), 'o-', label='AI Edge')
ax2.plot(df['Throughput (events/s)'], df['Legacy DPI RAM (MB)'].str.rstrip(' MB').astype(float), 's--', label='DPI Proxy')
ax2.set_title('RAM Footprint')
ax2.legend()
plt.tight_layout()
plt.show()
```
