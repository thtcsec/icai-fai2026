# Table III: Mean Time to Respond (MTTR) Analysis
This notebook reproduces Table III and Figure 2 comparing Manual SOC Triage vs Legacy SIEM vs AI-Native Autonomous Incident Response.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('table3_mttr.csv')
display(df)

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.bar(df['Operational Paradigm'], df['Mean MTTR (s)'].astype(float), color=['#d9534f', '#f0ad4e', '#5cb85c'])
ax.set_yscale('log')
ax.set_ylabel('MTTR (seconds)')
ax.set_title('MTTR Benchmark Comparison')
plt.tight_layout()
plt.show()
```
