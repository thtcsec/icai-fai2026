# Table IV: Anomaly Detection Performance & Threshold Sensitivity Analysis
This notebook reproduces Table IV and Figure 3 evaluating Precision, Recall, F1-Score, FPR, and FNR across reconstruction thresholds $\tau$.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('table4_precision.csv')
display(df)

fig, ax = plt.subplots(figsize=(6, 3.8))
ax.plot(df['Threshold (tau)'], df['Precision'], 'o-', label='Precision')
ax.plot(df['Threshold (tau)'], df['Recall'], 's-', label='Recall')
ax.plot(df['Threshold (tau)'], df['F1-Score'], '^--', label='F1-Score')
ax.set_xlabel('Threshold (tau)')
ax.set_ylabel('Score')
ax.legend()
plt.tight_layout()
plt.show()
```
